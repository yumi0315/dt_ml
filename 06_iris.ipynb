```python
import pandas as pd
from sklearn.tree import DecisionTreeClassifier
```

```python
iris = pd.read_csv('data/iris.csv')
# X: 입력값(꽃잎과 꽃받침의 길이와 넓이), 피처(Feature), 독립변수
# y: 출력값(꽃의 종류), 타겟(Target), 레이블(Label), 종속변수
X = iris.drop('variety', axis=1)
y = iris['variety']
```

```python
from sklearn.model_selection import train_test_split
from sklearn.model_selection import cross_val_score
from sklearn.ensemble import RandomForestClassifier
# 데이터 분할 (훈련 세트와 테스트 세트)
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=121)

# 모델 정의 및 훈련
clf = RandomForestClassifier(random_state=121)
clf.fit(X_train, y_train)
# 예측
pred = clf.predict(X_test)

clf2 = DecisionTreeClassifier(random_state=11)
clf2.fit(X_train,y_train)
pred2 = clf2.predict(X_test)

# 한줄에끝내기
scores = cross_val_score(clf, X, y, scoring='accuracy')
print(scores)
scores2 = cross_val_score(clf2, X, y, scoring='accuracy')
scores2
```

```python
from sklearn.metrics import accuracy_score
print('정확도: {}'.format(accuracy_score(y_test, pred)))
```

```python
print('정확도: {}'.format(accuracy_score(y_test, pred2)))
```

```python
# new_data = [[6., 2.9, 4.5, 1.5]]
new_data = pd.DataFrame([[6., 2.9, 4.5, 1.5]], columns=X.columns)
clf2.predict(new_data)
```
